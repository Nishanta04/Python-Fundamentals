# 14. Sets

```python
# Set is a collection of unordered items.
# Set ignores following any order and it also ignores duplicates
# Each elements in the set must be unique and immutable
# We can add and delete elements in set but cannot change the existing elements present in the set, because the elements in set are immutable.
# Type of values that can be stored in sets :
# 1. boolean
# 2. int
# 3. float
# 4. string
# 5. tuple
# dictionary and  list cannot be stored in sets because they are mutable.
collection = {1,23,6,4.5,"world", "hello", 1, 23}
print(collection)
print(collection)
```

```python
# creating an empty set
emp = set()
```

```python
emp.add(2)
emp.add(4)
print(emp)

emp.remove(2)
print(emp)

emp.clear()
print(emp)

print(collection.pop()) # pops(delete) out any random value from the set
print(collection)
```

```python
# Union and intersection in set
# Union and intersection in set returns a new set as output. It does not change the original
set1 = {1,2,3}
set2 = {3,4,5}

print(set1.union(set2))
print(set1.intersection(set2))
```

```python
n = int(input("Enter a number : "))
def fact(n):
  if(n == 0 or n == 1):
    return 1
  else:
    return n * fact(n-1)

print(fact(n))
```

# 1.  Objects and Classes

```python
# To map real world scenario, we started using objects in code. This is called Object oriented Programming.

# Creating Class
class Student:

  College_name = "SLIET"  # Class Attribute (Common For all objects in class)
                          # Accessing Class Attribute : Student.College_name

# Default Constructor :
  # def __init__(self):
  #   print("Creating new object")


# Parameterized Constructor :
  def __init__(self, name, marks):
    self.name = name      # Object Attribute (Different for every object)
    self.marks = marks    # Object Attribute > Class Attribute
    print("Creating new object")

# Self parameter is a reference to the current instance of the class, and is used to access variables that belongs to the class.
# creating object(instance)
s1 = Student("Nishanta", 92)
print(s1.name, s1.marks, s1.College_name)
print()

s2 = Student("Rahul", 89)
print(s2.name, s2.marks, Student.College_name)
print()
```

```python
# Object Attribute > Class Attribute
class Student2:
  College_name = "SLIET"
  name = "Anonymous"

  def __init__(self, name, marks):
    self.name = name
    self.marks = marks
    print("Creating new object")

s1 = Student2("Nishanta", 92)
print(s1.name, s1.marks, s1.College_name)
print()
```

```python
# Methods in class
class Student3:
  College_name = "SLIET"
  name = "Anonymous"

  def __init__(self, name, marks):
    self.name = name
    self.marks = marks

  def welcome(self):
    print("Welcome", self.name)

s1 = Student3("Nishanta", 92)
s1.welcome()
```

```python
# Create student class that takes name and marks of three subjects as arguments in constructor.
# Then create a method to print the average.

class Student4:

  def __init__(self, name, marks):
    self .name = name
    self.marks = marks

  def avg_marks(self):
    sum = 0
    for e in self.marks:
      sum += e
    print("Hey", self.name, "your average marks is : ", sum/3)


s4 = Student4("Nishanta", [34,45,49])
s4.avg_marks()
```
