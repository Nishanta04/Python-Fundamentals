```python
import re
import csv
import os

# ==========================================================
# Student Class
# ==========================================================

class Student:

    university = "Thapar Institute of Engineering and Technology"
    total_students = 0

    def __init__(self, name, age, roll_no, email):
        self.name = name
        self.age = age
        self.roll_no = roll_no
        self.__cgpa = 0

        if self.validate_email(email):
            self.email = email
        else:
            self.email = "Invalid Email"

        Student.total_students += 1

    # Email Validation
    def validate_email(self, email):
        pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'
        return re.match(pattern, email) is not None

    # Setter
    def set_cgpa(self, cgpa):
        if 0 <= cgpa <= 10:
            self.__cgpa = cgpa
        else:
            print("Invalid CGPA!")

    # Getter
    def get_cgpa(self):
        return self.__cgpa

    # Scholarship
    def calculate_scholarship(self):
        if self.__cgpa >= 9:
            return 50000
        elif self.__cgpa >= 8:
            return 25000
        return 0

# ==========================================================
# Function to Save Student to CSV
# ==========================================================

def save_to_csv(student):

    file_exists = os.path.isfile("students.csv")

    with open("students.csv", "a", newline="") as file:
        writer = csv.writer(file)

        # Write header only once
        if not file_exists:
            writer.writerow(["Name", "Age", "Roll No", "Email",
                             "CGPA", "Scholarship"])

        writer.writerow([
            student.name,
            student.age,
            student.roll_no,
            student.email,
            student.get_cgpa(),
            student.calculate_scholarship()
        ])

    print("Student data saved successfully!")

# ==========================================================
# Function to Display CSV Data
# ==========================================================

def display_csv():

    if not os.path.exists("students.csv"):
        print("No student records found.")
        return

    with open("students.csv", "r") as file:
        reader = csv.reader(file)

        print("\n========== STUDENT RECORDS ==========")
        for row in reader:
            print("\t".join(row))

# ==========================================================
# Main Menu
# ==========================================================

while True:

    print("\n========== STUDENT MANAGEMENT SYSTEM ==========")
    print("1. Add Student")
    print("2. Display All Students")
    print("3. Exit")

    choice = int(input("Enter your choice: "))

    if choice == 1:

        while True:

            print("\nEnter Student Details")

            name = input("Name: ")
            age = int(input("Age: "))
            roll = input("Roll Number: ")
            email = input("Email: ")
            cgpa = float(input("CGPA: "))

            student = Student(name, age, roll, email)
            student.set_cgpa(cgpa)

            save_to_csv(student)

            again = input("\nDo you want to add another student? (y/n): ")

            if again.lower() != 'y':
                break

    elif choice == 2:
        display_csv()

    elif choice == 3:
        print("Thank You!")
        break

    else:
        print("Invalid Choice!")
```
