# 1. What is Object-Oriented Programming?

Object-Oriented Programming (OOP) is a programming paradigm in which programs are organized around **objects** rather than functions.

An **object** contains:

- Data (Attributes)
- Functions (Methods)

Instead of writing one long program, we divide it into reusable classes.



# 2. Procedural Programming vs Object-Oriented Programming

| Feature | Procedural | Object-Oriented |
|----------|------------|----------------|
| Focus | Functions | Objects |
| Data | Separate | Encapsulated |
| Reusability | Low | High |
| Security | Low | High |
| Maintenance | Difficult | Easy |
| Real-world Modeling | Poor | Excellent |
| Example | C | Python, Java, C++ |

# 8. Methods

Methods define the behavior of an object.

```python

```



# 9. Encapsulation

Encapsulation hides internal data from direct access.

Private variables begin with **__**

Getter and Setter methods provide controlled access.



# 10. Abstraction

Abstraction hides implementation details.

The user only calls a function without knowing how it works.

Example:

```python
student.calculate_scholarship()
```



# 11. Inheritance

Inheritance allows one class to inherit properties from another.

```
          Student

```



# 12. Polymorphism

The same method behaves differently for different objects.

```

ResearchStudent -> I am a Research Student
SportsStudent -> I am a Sports Student
```

Implement the following methods:

1. **display()**
   - Display all the student's information in a readable format.

   - Assign the CGPA to the student.
   - Ensure that the CGPA is valid (between 0 and 10).

3. **calculate_scholarship()**
   - Calculate the scholarship amount based on the CGPA using the following criteria:
     - **CGPA ≥ 9.0:** ₹50,000
     - **7.0 ≤ CGPA < 9.0:** ₹25,000
     - **CGPA < 7.0:** ₹0

---



````python
# Define the base class named Student
class Student:

    # Class attribute shared by all Student objects
    university = "Thapar Institute of Engineering and Technology"

    # Constructor: automatically called when an object is created
    def __init__(self, name, age, roll_no):

        # Public attribute to store the student's name
        self.name = name

        # Public attribute to store the student's age
        self.age = age

        # Protected attribute (by convention) to store the roll number
        self._roll_no = roll_no

        # Private attribute to store CGPA (Encapsulation)
        self.__cgpa = 0

    # Method to display student details
    def display(self):

        # Print a blank line followed by the heading
        print("\nStudent Details")

        # Print a separator line
        print("----------------")

        # Display student's name
        print("Name :", self.name)

        # Display student's age
        print("Age  :", self.age)

        # Display student's roll number
        print("Roll :", self._roll_no)

        # Display the university name using the class attribute
        print("University:", Student.university)

    # Setter method to assign CGPA
    def set_cgpa(self, cgpa):

        # Validate that CGPA lies between 0 and 10
        if 0 <= cgpa <= 10:

            # Store the valid CGPA in the private attribute
            self.__cgpa = cgpa

    # Getter method to access the private CGPA
    def get_cgpa(self):

        # Return the current CGPA
        return self.__cgpa

    # Method to calculate scholarship amount
    def calculate_scholarship(self):

        # If CGPA is 9 or above
        if self.__cgpa >= 9:

            # Return scholarship amount
            return 50000

        # If CGPA is between 8 and 9
        elif self.__cgpa >= 8:

            # Return scholarship amount
            return 25000

        # Otherwise no scholarship
        return 0

    # Method to introduce the student
    def introduce(self):

        # Display the type of object
        print("I am a Student.")
````
