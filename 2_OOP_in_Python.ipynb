```python
# Import the regular expression module
# Used for validating email addresses
import re

# ==========================================================
# Base Class
# ==========================================================

# Define a class named Student
class Student:

    # Class attribute shared by all Student objects
    university = "Thapar Institute of Engineering and Technology"

    # Class attribute to count the total number of students
    total_students = 0

    # Constructor: Automatically executed whenever an object is created
    def __init__(self, name, age, roll_no, email):

        # Public attribute to store the student's name
        self.name = name

        # Public attribute to store the student's age
        self.age = age

        # Public attribute to store the student's roll number
        self.roll_no = roll_no

        # Private attribute (Encapsulation)
        # Can only be accessed through getter and setter methods
        self.__cgpa = 0

        # Validate the email address before storing it
        if self.validate_email(email):

            # Store the email if it is valid
            self.email = email

        else:

            # Store an error message if the email is invalid
            self.email = "Invalid Email"

        # Increase the total student count whenever a new object is created
        Student.total_students += 1


    # ------------------------------------------------------
    # Email Validation
    # ------------------------------------------------------

    # Method to validate the email format
    def validate_email(self, email):

        # Regular expression for email validation
        pattern = r'^[A-Za-z0-9._%+-]+@[A-Za-z0-9.-]+\.[A-Za-z]{2,}$'

        # Return True if email matches the pattern; otherwise False
        return re.match(pattern, email) is not None


    # ------------------------------------------------------
    # Setter Method
    # ------------------------------------------------------

    # Method to assign a CGPA
    def set_cgpa(self, cgpa):

        # Check whether the CGPA is valid
        if 0 <= cgpa <= 10:

            # Store the CGPA in the private variable
            self.__cgpa = cgpa

        else:

            # Display an error message for invalid input
            print("Invalid CGPA")


    # ------------------------------------------------------
    # Getter Method
    # ------------------------------------------------------

    # Method to retrieve the private CGPA
    def get_cgpa(self):

        # Return the stored CGPA
        return self.__cgpa


    # ------------------------------------------------------
    # Abstraction Example
    # ------------------------------------------------------

    # Method to calculate scholarship amount
    def calculate_scholarship(self):

        # If CGPA is 9 or above
        if self.__cgpa >= 9:

            # Return scholarship amount
            return 50000

        # If CGPA is between 8 and 9
        elif self.__cgpa >= 8:

            # Return scholarship amount
            return 25000

        # Otherwise no scholarship
        return 0


    # ------------------------------------------------------
    # Display Method
    # ------------------------------------------------------

    # Method to display student details
    def display(self):

        # Print the object using the __str__() method
        print(self)

        # Display the CGPA
        print("CGPA         :", self.get_cgpa())

        # Display the scholarship amount
        print("Scholarship  :", self.calculate_scholarship())


    # ------------------------------------------------------
    # Method for Polymorphism
    # ------------------------------------------------------

    # Method that will be overridden in child classes
    def introduce(self):

        # Display the object type
        print("I am a Student.")


    # ------------------------------------------------------
    # __str__() Method
    # ------------------------------------------------------

    # Returns a formatted string representation of the object
    def __str__(self):

        return f"""
-------------------------------
Name       : {self.name}
Age        : {self.age}
Roll No    : {self.roll_no}
Email      : {self.email}
University : {Student.university}
-------------------------------
"""


# ==========================================================
# Child Class
# Demonstrates Inheritance
# ==========================================================

# ResearchStudent inherits all properties of Student
class ResearchStudent(Student):

    # Constructor of the child class
    def __init__(self, name, age, roll_no, email, research_area):

        # Call the constructor of the parent class
        super().__init__(name, age, roll_no, email)

        # New attribute specific to ResearchStudent
        self.research_area = research_area

    # Method specific to ResearchStudent
    def research(self):

        # Display research information
        print(self.name, "is working in", self.research_area)

    # Method overriding (Polymorphism)
    def introduce(self):

        # Display the object type
        print("I am a Research Student.")



# ==========================================================
# Another Child Class
# ==========================================================

# SportsStudent inherits from Student
class SportsStudent(Student):

    # Constructor
    def __init__(self, name, age, roll_no, email, sport):

        # Call the parent constructor
        super().__init__(name, age, roll_no, email)

        # Store the sport played
        self.sport = sport

    # Method specific to SportsStudent
    def play(self):

        # Display sport information
        print(self.name, "plays", self.sport)

    # Method overriding (Polymorphism)
    def introduce(self):

        # Display the object type
        print("I am a Sports Student.")
```

```python
# ==========================================================
# Creating Objects
# ==========================================================
# Create an object of the Student class
student1 = Student(
    "Amit",
    21,
    "1023",
    "amit@thapar.edu"
)

```

```python
# ==========================================================
# Demonstrate Polymorphism
# ==========================================================

# Display a heading
print("\nPolymorphism Demonstration")

# Store objects of different classes in a single list
students = [student1, student2, student3]

# Loop through each object
for student in students:
    # Same method call produces different outputs
    # depending on the object's class
    student.introduce()
```

```python
# ==========================================================
# Display Total Number of Students
# ==========================================================
# Print the total number of Student objects created
print("\nTotal Students =", Student.total_students)
```

```python
# ==========================================================
# Sort Students by CGPA
# ==========================================================
# Sort the list in descending order of CGPA
students.sort(
    key=lambda s: s.get_cgpa(),
    reverse=True
)

# Display the heading
print("\nStudents Sorted by CGPA")

# Display each student's name and CGPA
for student in students:

    print(student.name, ":", student.get_cgpa())
```

## Summary Table

| **Concept** | **Definition** | **Python Example** |
|--------------|----------------|--------------------|
| **Class** | Blueprint for creating objects | `class Student:` |
| **Object** | Instance of a class | `student1 = Student()` |
| **Constructor** | Initializes object attributes automatically | `__init__()` |
| **Attributes** | Variables that store object data | `self.name`, `self.age` |
| **Methods** | Functions defined inside a class | `display()` |
| **Encapsulation** | Hides data using private/protected attributes | `__cgpa`, `set_cgpa()` |
| **Abstraction** | Hides implementation details and exposes only essential functionality | `calculate_scholarship()` |
| **Inheritance** | Child class acquires properties and methods of a parent class | `class ResearchStudent(Student):` |
| **Polymorphism** | Same method behaves differently for different objects | Overridden `introduce()` method |
